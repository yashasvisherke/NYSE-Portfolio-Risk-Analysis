```python
# --- 1. Load the data ---
df = pd.read_csv('../input/nyse/prices-split-adjusted.csv', parse_dates=['date'])

# --- 2. Calculate daily returns for all stocks ---
df['daily_return'] = df.groupby('symbol')['close'].pct_change()

# --- 3. Create the "wide" DataFrame of returns ---
returns_df = df.pivot_table(index='date', columns='symbol', values='daily_return')

# --- 4. Create the correlation matrix ---
correlation_matrix = returns_df.corr()

# --- 5. Now, get the specific value ---
# (We put this in a "try...except" block to be safe)
try:
    aapl_msft_corr = correlation_matrix.loc['AAPL', 'MSFT']
    print(aapl_msft_corr)
except KeyError as e:
    print(f"Error: Could not find key {e}. Check your symbols.")
```

```python
# 1. Get the list of correlations for just AAPL
aapl_correlations = correlation_matrix.loc['AAPL']

# 2. Sort that list from smallest to largest
best_hedges = aapl_correlations.sort_values()

# 3. Print the top 5 best hedges
print(best_hedges.head())
```

```python
# --- 1. Define your weights ---
w_aapl = 0.5
w_evhc = 0.5

# --- 2. Get your three calculated values ---
# (You might need to re-run the cells that define
# 'all_volatility' and 'correlation_matrix' first!)
try:
    vol_aapl = all_volatility.loc['AAPL']
    vol_evhc = all_volatility.loc['EVHC']
    corr_aapl_evhc = correlation_matrix.loc['AAPL', 'EVHC']

    print(f"AAPL Volatility (σ1): {vol_aapl}")
    print(f"EVHC Volatility (σ2): {vol_evhc}")
    print(f"Correlation (p1,2): {corr_aapl_evhc}")

except NameError:
    print("Error: 'all_volatility' or 'correlation_matrix' not defined.")
    print("Please re-run your previous cells to define them.")
except KeyError as e:
    print(f"Error: Could not find key {e}. Did all your previous cells run?")
```

```python
# Term 1: (w_aapl^2 * vol_aapl^2)
term1 = (w_aapl**2) * (vol_aapl**2)

# Term 2: (w_evhc^2 * vol_evhc^2)
term2 = (w_evhc**2) * (vol_evhc**2)

# Term 3: (2 * w_aapl * w_evhc * correlation * vol_aapl * vol_evhc)
term3 = 2 * w_aapl * w_evhc * corr_aapl_evhc * vol_aapl * vol_evhc

# --- 1. Calculate the Portfolio Variance ---
portfolio_variance = term1 + term2 + term3

# --- 2. Calculate the Portfolio Volatility ---
# (This is the final answer: the square root of variance)
portfolio_volatility = portfolio_variance ** 0.5


print(f"Portfolio Variance: {portfolio_variance}")
print(f"Portfolio Volatility (Risk): {portfolio_volatility}")
```
